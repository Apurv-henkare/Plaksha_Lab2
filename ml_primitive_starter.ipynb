```python
import numpy as np
from knn import plot_knn_prediction
from gradient_descent import plot_gradient_descent,plot_loss_function,gradient_descent_2d,plot_2d_gradient_descent
from pca import pca_via_svd,plot_pca_reconstruction
```

```python
np.random.seed(0)
class_a = np.random.randn(30, 2) + np.array([2, 2])
#print ("Class a:",class_a)
class_b = np.random.randn(30, 2) + np.array([-2, -2])
#print ("Class b:",class_b)
X_train = np.vstack([class_a, class_b])
y_train = np.array([0] * 30 + [1] * 30)
```

```python
plot_knn_prediction(X_train,y_train,3,[2.5, 3.1])
```

```python
plot_gradient_descent(0.8,0.5,0.1)
```

```python
plot_loss_function(0.01,0.1,1.1)
```

```python
my_lr = 0.08
my_steps = 20

# Calculate
my_history = gradient_descent_2d(start_point=[3.0, 2.0], lr=my_lr, steps=my_steps)

# Plot
plot_2d_gradient_descent(my_history, lr=my_lr, steps=my_steps)
```

```python
# Set a seed so the "random" numbers are the same every time you run it
np.random.seed(42)

# 1. Generate 100 random x values
x_pca = np.random.randn(100)

# 2. Compute y = 2x + some random noise
noise = np.random.randn(100) * 0.5
y_pca = 2 * x_pca + noise

# 3. Stack and transpose to get our (100, 2) dataset
X_pca = np.vstack([x_pca, y_pca]).T
```

```python
projected, pc, mean = pca_via_svd(X_pca, n_components=1)

# 2. Call our new clean plotting function!
plot_pca_reconstruction(X_pca, projected, pc, mean)
```
